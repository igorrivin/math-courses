# Distributions and moments
Move among probability mass, simulation, expectation, and approximation.

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(2026)
x = rng.binomial(20, .3, size=10000)
values, counts = np.unique(x, return_counts=True)
plt.bar(values, counts/counts.sum(), color='#6750a4')
plt.xlabel('successes'); plt.ylabel('empirical probability'); plt.show()
print('empirical mean/variance:', x.mean(), x.var())
print('theoretical mean/variance:', 20*.3, 20*.3*.7)
```

## Rare-event approximation
Compare Binomial$(n,p)$ with Poisson$(np)$ while changing $n$ and $p$ but holding $np$ fixed.

```python
from math import comb, exp, factorial
n, p = 1000, .002
lam = n*p
for k in range(7):
    b = comb(n,k)*p**k*(1-p)**(n-k)
    q = exp(-lam)*lam**k/factorial(k)
    print(k, f'{b:.6f}', f'{q:.6f}', f'error={abs(b-q):.2e}')
```
