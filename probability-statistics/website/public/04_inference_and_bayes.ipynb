# Confidence, testing, and Bayesian updating
Compare repeated-sampling coverage with posterior probability.

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(2026)
p, n, runs = .55, 100, 5000
phat = rng.binomial(n,p,size=runs)/n
se = np.sqrt(phat*(1-phat)/n)
covered = (phat-1.96*se <= p) & (p <= phat+1.96*se)
print('empirical 95% coverage:', covered.mean())
```

## Beta-binomial update
Treat alpha and beta as adjustable prior pseudocounts.

```python
alpha, beta = 2, 2
successes, failures = 8, 2
post_a, post_b = alpha+successes, beta+failures
draws = rng.beta(post_a, post_b, 200000)
print('posterior mean:', draws.mean())
print('95% credible interval:', np.quantile(draws,[.025,.975]))
x = np.linspace(0,1,400)
plt.hist(draws,bins=70,density=True,color='#6750a4',alpha=.65); plt.xlabel('p'); plt.show()
```
