# Lab 3: Pascal's triangle sees primes
Move from divisibility pictures to Chebyshev's prime-counting scale.

```sage
def pascal_mod(p, rows):
    return matrix(ZZ, rows, rows,
                  lambda n,k: binomial(n,k) % p if k <= n else 0)

matrix_plot(pascal_mod(2, 64), cmap='binary', colorbar=False,
            axes=False, frame=False, origin='upper')
```

Repeat for $p=3$ and $p=5$. Formulate a base-$p$ digit rule for the zero entries. Relate it to carries in $k+(n-k)$.

```sage
def central_squeeze(n):
    c = binomial(2*n,n)
    return (4^n/(2*n+1), c, 4^n)

[central_squeeze(n) for n in [5,10,25]]
```

```sage
def theta(x):
    return sum(log(RR(p)) for p in prime_range(floor(x)+1))

def psi(x):
    total = 0
    for p in prime_range(floor(x)+1):
        pk = p
        while pk <= x:
            total += log(RR(p))
            pk *= p
    return total

[(x, prime_pi(x), theta(x), psi(x)) for x in [10,100,1000]]
```

```sage
xs = list(range(3,10001,10))
P = list_plot([(x, prime_pi(x)) for x in xs], plotjoined=True,
              legend_label=r'$\pi(x)$', color='navy')
A = plot(x/log(x), (x,3,10000), color='darkgreen',
         legend_label=r'$x/\log x$')
L = plot((log(2)/2)*x/log(x), (x,3,10000), color='orange',
         legend_label='Chebyshev lower')
U = plot(4*x/log(x), (x,3,10000), color='red',
         legend_label='Chebyshev upper')
P + A + L + U
```
